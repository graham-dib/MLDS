The discrepancy here is $T(\y,\beta)=\sum_i (y_i - (X\beta)_i)^2$ — the residual sum of
squares. Look up Definition 7.4.2 in your notes before starting.

## The one structural thing to get right

$T(\y,\beta)$ depends on **$\beta$ as well as the data**. So it is a *discrepancy*, not a
test statistic: unlike a classical $\chi^2$, the "observed" value is itself random, because
$\beta$ is random. That has a consequence for how the comparison must be done.

**Q1.** Below you are handed *paired* arrays — the $k$th entry of each was computed at the
same posterior draw $\beta^{(k)}$. Why does the pairing matter? What would break if you
compared, say, the sorted values instead?

```python
import numpy as np

rng = np.random.default_rng(0)
N = 20_000


def ppp(T_obs, T_rep):
    """Given paired discrepancies, the p-value is just a proportion."""
    return np.mean(T_rep >= T_obs)


# Pretend someone else generated these for you (they are not from xy.txt).
T_rep = rng.chisquare(df=8, size=N)
T_obs = rng.chisquare(df=8, size=N)
print(f"p-hat = {ppp(T_obs, T_rep):.4f}")
```

## What the value tells you

Three regimes, same machinery.

```python
for label, scale in (("data as dispersed as the model expects", 1.00),
                     ("data much LESS scattered than the model expects", 0.35),
                     ("data much MORE scattered than the model expects", 2.50)):
    T_rep = rng.chisquare(df=8, size=N)
    T_obs = scale * rng.chisquare(df=8, size=N)
    print(f"p-hat = {ppp(T_obs, T_rep):.4f}   {label}")
```

## Monte Carlo error

$\hat p$ is the mean of $N$ indicator variables, so its standard error is available in closed
form. Worth a clause in your answer — it is what tells you whether $N$ was large enough.

```python
for p_hat in (0.50, 0.92, 0.999):
    for N_draws in (1_000, 20_000):
        se = np.sqrt(p_hat * (1 - p_hat) / N_draws)
        print(f"p-hat = {p_hat:5.3f}, N = {N_draws:6d}  ->  s.e. = {se:.5f}")
    print()
```
