## The mechanism, in three lines

Posterior model probabilities are *normalised over the candidate set*. Watch what that does
when every candidate is dreadful.

```python
import numpy as np

# Three catastrophically bad models -- log marginal likelihoods of -400 or so.
log_ml = np.array([-431.0, -428.0, -420.0])   # M1, M2, M3

post = np.exp(log_ml - log_ml.max())
post /= post.sum()

for i, p in enumerate(post, start=1):
    print(f"P(M{i} | y) = {p:.6f}")
print(f"\nsum = {post.sum():.6f}")
print("\nM3 looks overwhelmingly preferred. Nothing here knows the models are terrible.")
```

**Q1.** The probabilities are forced to sum to 1 whatever the absolute values are. Say what
kind of question posterior model probabilities answer, and what kind they *cannot* answer.
Two words that may help: **relative** vs **absolute**.

**Q2.** A posterior predictive check does not normalise over anything — it compares one model
against the data alone. Why does that make it capable of detecting something the model
probabilities structurally cannot?

## Same point, second angle

Introduce a fourth model that was never in the running and re-normalise.

```python
log_ml4 = np.append(log_ml, -390.0)   # M4: a genuinely better model, e.g. a different error law

post4 = np.exp(log_ml4 - log_ml4.max())
post4 /= post4.sum()

for i, (p_old, p_new) in enumerate(zip(np.append(post, np.nan), post4), start=1):
    old = "     --   " if np.isnan(p_old) else f"{p_old:10.3e}"
    print(f"P(M{i} | y):  before {old}   after {p_new:10.3e}")

print("\nM3's 'overwhelming' support evaporated -- the data never changed.")
print(f"\nBut the Bayes factor B_32 is untouched: "
      f"before {np.exp(log_ml[2] - log_ml[1]):.4f}, after {np.exp(log_ml4[2] - log_ml4[1]):.4f}")
```

**Q3.** Note the last line: every posterior model probability moved, but $B_{32}$ did not. The
Bayes factors between the original three carry the same evidence as before — only the
normalisation shifted. Tie this back to what you wrote in part (a) about $B_{ij}$ being free of
the prior model weights.
